# Exploratory Data Analysis: Student Pass/Fail Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Configure visualization aesthetics
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

DATA_PATH = os.path.join("..", "data", "student_data.csv")
# Load data
df = pd.read_csv(DATA_PATH)
print(f"Dataset shape: {df.shape}")
df.head()
```

## 1. Class Distribution

```python
# Pass/Fail Distribution
sns.countplot(data=df, x="pass_fail", palette="Set2")
plt.title("Pass vs Fail Distribution")
plt.show()
```

## 2. Feature Distributions

```python
features = [
    "daily_study_hours", "attendance_rate", "sleep_hours", 
    "stress_level", "motivation_score", "math_score", 
    "reading_score", "writing_score"
]

fig, axes = plt.subplots(2, 4, figsize=(20, 10))
axes = axes.flatten()

for i, feature in enumerate(features):
    sns.histplot(df[feature], bins=20, kde=True, ax=axes[i], color="skyblue")
    axes[i].set_title(f"Distribution of {feature}")
    
plt.tight_layout()
plt.show()
```

## 3. Features by Outcome (Pass/Fail) - Boxplots & Violin Plots

```python
# Boxplots show median and quartiles
fig, axes = plt.subplots(2, 4, figsize=(20, 10))
axes = axes.flatten()

for i, feature in enumerate(features):
    sns.boxplot(data=df, x="pass_fail", y=feature, palette="Set2", ax=axes[i])
    axes[i].set_title(f"{feature} (Boxplot)")

plt.tight_layout()
plt.show()
```

```python
# Violin plots show full data density
fig, axes = plt.subplots(2, 4, figsize=(20, 10))
axes = axes.flatten()

for i, feature in enumerate(features):
    sns.violinplot(data=df, x="pass_fail", y=feature, palette="pastel", split=True, ax=axes[i], inner="quart")
    axes[i].set_title(f"{feature} (Violin)")

plt.tight_layout()
plt.show()
```

## 4. Key Target Features Pairplot

```python
# Sampling data to avoid lag with massive datasets
sample_df = df.sample(n=min(5000, len(df)), random_state=42)
key_features = ["math_score", "reading_score", "writing_score", "attendance_rate", "pass_fail"]

sns.pairplot(sample_df[key_features], hue="pass_fail", palette="Set2", corner=True, plot_kws={"alpha": 0.6})
plt.suptitle("Pairplot of Key Academic Scores vs Attendance", y=1.02)
plt.show()
```

## 5. Correlation Matrix

```python
# Encode pass_fail temporarily for correlation
df["pass_fail_numeric"] = df["pass_fail"].apply(lambda x: 1 if x == "Pass" else 0)

numeric_cols = df.select_dtypes(include=["number"])
corr = numeric_cols.corr()

plt.figure(figsize=(10, 8))
# Mask the upper triangle for a cleaner look
import numpy as np
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix of Numeric Features")
plt.show()

# Clean up
df.drop(columns=["pass_fail_numeric"], inplace=True)
```
